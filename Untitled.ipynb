### groupby()
- Used to Split Data into groups based on the values of one or more column.
- After Splitting . it performs Aggregation on each group seperately.
- ### Agg Functions like
- sum()
- mean()/avg()
- count()
- max()
- min()
- median()

```python
data = {
    'EmpID':[1,2,3,4,5,6],
    'Name':['A','B','C','D','E','F'],
    'Department':['Sales','HR','Sales','IT','HR','IT'],
    'salary':[50000,60000,50000,70000,60000,70000]
}
df = pd.DataFrame(data)
df
```

```python
summary = df.groupby('Department').agg(
    Avg_salary = ('salary','mean'),
    Total_Salary = ('salary','sum'),
    Employee_count = ('EmpID','count')
)
summary
```

```python
#find Maximum salary per department
max_salary = df.groupby('Department')['salary'].max()
max_salary
```

```python
min_salary = df.groupby('Department')['salary'].min()
min_salary
```

```python
multi = df.groupby('Department')['salary'].agg(
    ['mean','sum','min','max','count','median']
)
multi
```

```python
data = {
    'OrderID':[1,2,3,4,5,6],
    'Region':['east','west','east','south','west','east'],
    'Product':['laptop','mobile','laptop','tablet','laptop','mobile'],
    'Sales':[800,500,900,700,1200,400]
}
df = pd.DataFrame(data)
df

#total sales by Region
#Average sales by product
#count of orders per region
```

```python
summary = df.groupby('Region').agg(
    Total_sales = ('Sales','sum'),
    Average_sales = ('Sales','mean'),
    count_orders = ('OrderID','count')
)
summary
```

```python
total_sales_region = df.groupby('Region')['Sales'].sum()
total_sales_region

```

```python
avg_sales_product = df.groupby('Product')['Sales'].mean()
avg_sales_product
```

```python
order_count_region = df.groupby('Region')['OrderID'].count()
order_count_region 
```

### Categorical analysis using crosstab
- use on two or more categorical variables.
- cross tabulation -> Contingency table
- shows how categories of one variable relate to categories of another.
- by default , it counts occurrences.
-  we can also summarize numeric data 

```python
data = {
    'Employee':['John','James','David','Mary','Tom','Charlie'],
    'Department':['Sales','HR','Sales','IT','HR','IT'],
    'Gender':['Male','Female','Male','Female','Male','Male'],
    'Salary':[50000,60000,55000,70000,62000,68000]
}
df = pd.DataFrame(data)
df
```

```python
ct = pd.crosstab(index = df['Department'],columns = df['Gender'])
ct
```

```python
ct_sum = pd.crosstab(index = df['Department'],columns = df['Gender'],
                     values = df['Salary'],aggfunc = 'sum'
                    )
ct_sum
```

```python
ct_fill = pd.crosstab(index = df['Department'],columns = df['Gender'],
                     values = df['Salary'],aggfunc = ['sum','mean'],
                    )
ct_fill
```

```python
ct_margins = pd.crosstab(index = df['Department'],columns = df['Gender'],
                     values = df['Salary'],aggfunc = 'sum', margins = True
                    )
ct_margins
```

```python
ct_norm = pd.crosstab(index = df['Department'],columns = df['Gender'],normalize = 'index')
ct_norm
```

### Concatenation

```python
data_a = {
    'EmpID':[1,2],
    'Name':['John','Tom'],
    'Department':['Sales','HR']
}

data_b = {
    'EmpID':[3,24],
    'Name':['Dharam','veer'],
    'Department':['IT','Finance']
}

df_a = pd.DataFrame(data_a)
df_b = pd.DataFrame(data_b)
print(df_a)
print('-------------------------')
print(df_b)
```

```python
#row wise concatenation

row_concat = pd.concat([df_a,df_b], axis = 0)
row_concat
```

```python
row_concat = pd.concat([df_a,df_b],axis = 0,ignore_index = True)
row_concat
```

```python
col_concat = pd.concat([df_a,df_b],axis = 1)
col_concat
```

### Merging

```python
employees = pd.DataFrame({
    'EmpID':[1,2,3,4],
    'EmpName':['A','B','C','D'],
    'DeptID':[101,102,103,104]
})

departments = pd.DataFrame({
    'DeptID':[101,102,103,105],
    'DeptName':['HR','IT','Finance','Marketing']
})
print(employees)
print('-----------------------')
print(departments)
```

```python
#inner join --> returns common rows only
inner = pd.merge(employees,departments,on = 'DeptID',how = 'inner')
print(inner)
```

```python
#left join --> all Employees + matching departments
left = pd.merge(employees,departments,on = 'DeptID',how = 'left')
print(left)
```

```python
#right join --> all Employees + matching emp
right = pd.merge(employees,departments,on = 'DeptID',how = 'right')
print(right)
```

```python
#outer join --> All record from both DataFrame
outer = pd.merge(employees,departments,on = 'DeptID',how = 'outer')
print(outer)
```
