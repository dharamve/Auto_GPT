### aplly()
- Applies a function on rows and coulumns of Df and each element  of Series.
- DataFrame or series method
  

```python
import pandas as pd
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
data = {
    'Name':['Arun','Bishop','Capher','Dharam'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Joining_year':[2019,2020,2018,2021]
}

df = pd.DataFrame(data)
df
```

```python
df['Updated_Salaries'] = df['Salary'].apply(lambda x:x+1000)
df
```

```python
def salary_category(salary):
    if salary >= 60000:
        return 'High'
    else:
        return 'Low'
df['Salary_category'] = df['Salary'].apply(salary_category)
df
```

```python
def salary_hike(Updated_Salaries):
    return Updated_Salaries + 5000
df['salary_hike'] = df['Updated_Salaries'].apply(salary_hike)
df
    
```

```python
import datetime

current = datetime.datetime.now().year
df['Experience'] = df.apply(lambda row: current - row['Joining_year'],axis = 1)
df
```

### Map()

```python
data = {
    'Name':['Arun','Bishop','Capher','Dharam'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Joining_year':[2019,2020,2018,2021]
}

df = pd.DataFrame(data)
df
```

```python
df['Updated_Salary'] = df['Salary'].map(lambda x:x+5000)
df
```

```python
dept_map = {'HR':'D1','IT':'D2','Finance':'D3'}
df['Dept_code'] = df['Department'].map(dept_map)
df
```

### Applymap()
- DataFrame method used to apply a function element wise to every single value in a DataFrame.
- Always Word on individual elemtn of DataFrame , not on Series

```python
data = {
    'Name':['Arun','Bishop','Capher','Dharam'],
    'Department':['HR','IT','Finance','IT'],
    'Salary':[50000,60000,55000,70000],
    'Bonus':[5000,4000,5500,6000]
}

df = pd.DataFrame(data)
df
```

```python
def add_numeric(x):
    if isinstance(x,(int,float)):
        return x + 1000
    return x
df_numeric_add = df.applymap(add_numeric)
df_numeric_add
```

```python
def text_upper(x):
    if isinstance(x,str):
        return x.upper()
    return x
df_text_update = df.applymap(text_upper)
df_text_update
```

```python
df_upper = df.applymap(lambda x:x.upper() if isinstance(x,str) else x)
df_upper
```

```python
def high_salary(x):
    if isinstance(x,(int,float)) and x > 60000:
        return 'High'
    return x
df3 = df.applymap(high_salary)
df3
```

### Window Functions
- We Calculate on a Small Part of the dataset.
- Two Types
- rolling()
- expanding()
- ##### Rolling()
- Creates a moving / sliding window
- Specify the no of rows
- apply agg Function on that rows
- ##### Expanding()
- Creates a growing windoe from the first element.
- Each row includes all previous rows in the calculation.
- cummulative Avg , sum etc

```python
sales = pd.Series([10,20,30,40,50])

sales
```

```python
print("Rollig:\n",sales.rolling(3).mean())
print('----------------------')
print("Expanding:\n",sales.expanding().mean())
```
